```python
product_ids = ["P101", "P102", "P103", "P101", "P102", "P104"]
def find_duplicate(dup):
    duplicate=[]
    for k,i in enumerate(dup):
        for j in range(k+1,len(dup)):
            if i not in duplicate and i==dup[j]:
                duplicate.append(i)
    return duplicate              
print(find_duplicate(product_ids))
```
